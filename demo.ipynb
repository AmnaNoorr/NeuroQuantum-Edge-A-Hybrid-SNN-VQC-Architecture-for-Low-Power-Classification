# NeuroQuantum-Edge: results demo

[![Open In Colab](https://colab.research.google.com/assets/colab-badge.svg)](https://colab.research.google.com/github/AmnaNoorr/NeuroQuantum-Edge-A-Hybrid-SNN-VQC-Architecture-for-Low-Power-Classification/blob/main/notebooks/demo.ipynb)

Three binary ECG classifiers (140 samples per heartbeat) are compared on the same stratified 70/15/15 split:

- **MLP**: classical baseline
- **SNN**: pure spiking neural network (snnTorch)
- **Hybrid SNN+VQC**: spiking feature extractor, a linear bottleneck to a few qubits, and a PennyLane variational quantum circuit (simulated)

Everything below is read from the saved `results/*/seed_*.json` files. **No training happens in this notebook**, and smoke-test runs are ignored automatically.

To reproduce the numbers, run `python -m src.train --config configs/<name>.yaml --seed <n>` for seeds 0-4, then re-run this notebook.

```python
import json, os, subprocess, sys
from pathlib import Path

REPO_URL = "https://github.com/AmnaNoorr/NeuroQuantum-Edge-A-Hybrid-SNN-VQC-Architecture-for-Low-Power-Classification.git"


def find_repo_root():
    here = Path.cwd().resolve()
    for p in [here, *here.parents]:
        if (p / "results").is_dir() and (p / "src").is_dir():
            return p
    return None


root = find_repo_root()
if root is None and "google.colab" in sys.modules:
    subprocess.run(["git", "clone", "-q", REPO_URL, "nqe"], check=True)
    root = Path("nqe").resolve()
if root is None:
    raise RuntimeError("Run this notebook from inside the repository (it needs the results/ and src/ folders).")
os.chdir(root)
print("Repository root:", root)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

plt.rcParams.update({
    "figure.dpi": 110,
    "axes.grid": True,
    "grid.alpha": 0.3,
    "axes.spines.top": False,
    "axes.spines.right": False,
})

NAMES = {"classical_mlp": "MLP", "pure_snn": "SNN", "hybrid_snn_vqc": "Hybrid SNN+VQC"}
COLORS = {"MLP": "#4C72B0", "SNN": "#DD8452", "Hybrid SNN+VQC": "#55A868"}

# Which CSV label is the normal heartbeat? In the common Kaggle version of this dataset the
# smaller class (about 41.6 percent of rows) is label 0 = abnormal, and label 1 = normal.
# Confirm with the ECG plot at the bottom of this notebook and change this if it is reversed.
NORMAL_LABEL = 1
ABNORMAL_LABEL = 1 - NORMAL_LABEL
```

## 1. Load results

```python
records, skipped = [], []
for path in sorted(Path("results").glob("*/seed_*.json")):
    run = json.loads(path.read_text(encoding="utf-8"))
    if run.get("smoke"):
        skipped.append(path.as_posix())
        continue
    test = run["test"]
    cm = np.array(test["confusion_matrix"])  # rows = true label, columns = predicted label
    a, n = ABNORMAL_LABEL, NORMAL_LABEL
    tp, fn, fp, tn = cm[a, a], cm[a, n], cm[n, a], cm[n, n]
    records.append({
        "experiment": run["experiment"],
        "model": NAMES.get(run["experiment"], run["experiment"]),
        "seed": run["seed"],
        "device": run.get("device", "?"),
        "accuracy": test["accuracy"],
        "macro_f1": test["macro_f1"],
        "auroc": test["auroc"],
        "abnormal_recall": tp / (tp + fn) if (tp + fn) else float("nan"),
        "abnormal_precision": tp / (tp + fp) if (tp + fp) else float("nan"),
        "specificity": tn / (tn + fp) if (tn + fp) else float("nan"),
        "params": test["parameter_count"],
        "ms_per_sample": test["inference_time_ms_per_sample"],
        "best_epoch": run.get("best_epoch"),
        "epochs": run.get("epochs_completed"),
        "cm": cm,
        "history": run.get("history", []),
    })

df = pd.DataFrame(records)
if df.empty:
    raise RuntimeError("No full (non-smoke) result files found in results/. Run the experiments first.")

ORDER = [m for m in NAMES.values() if m in set(df["model"])] + sorted(set(df["model"]) - set(NAMES.values()))
_palette = plt.get_cmap("tab10")


def color_of(model):
    return COLORS.get(model, _palette(ORDER.index(model) % 10))


print(f"Loaded {len(df)} full runs; ignored {len(skipped)} smoke-test file(s).")
```

```python
# Sanity checks: catch stale or missing runs before trusting any number below.
seed_counts = df.groupby("model")["seed"].nunique().rename("seeds").to_frame()
display(seed_counts.loc[ORDER])
for model in ORDER:
    g = df[df["model"] == model]
    if g["params"].nunique() > 1:
        print(f"WARNING: {model} has runs with different parameter counts {sorted(g['params'].unique())}. "
              "Stale results from an older version of the code are probably mixed in; re-run that model.")
    if g["seed"].nunique() < 3:
        print(f"NOTE: {model} has fewer than 3 seeds, so the std is not very meaningful yet.")
print("Devices used:", sorted(df["device"].unique()))
```

## 2. Test-set metrics (mean ± std over seeds)

```python
METRICS = {
    "accuracy": "Accuracy",
    "macro_f1": "Macro F1",
    "auroc": "AUROC",
    "abnormal_recall": "Abnormal recall",
    "abnormal_precision": "Abnormal precision",
    "specificity": "Specificity",
}


def pm(values, digits=4):
    values = pd.Series(values).dropna()
    std = values.std() if len(values) > 1 else 0.0
    return f"{values.mean():.{digits}f} ± {std:.{digits}f}"


rows = []
for model in ORDER:
    g = df[df["model"] == model]
    row = {"Model": model, "Seeds": len(g)}
    for col, label in METRICS.items():
        row[label] = pm(g[col])
    row["Parameters"] = f"{g['params'].mean():,.0f}"
    row["Inference (ms/sample)"] = pm(g["ms_per_sample"], 3)
    rows.append(row)

summary = pd.DataFrame(rows).set_index("Model")
display(summary)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
rng = np.random.default_rng(0)
for ax, (col, label) in zip(axes, [("macro_f1", "Macro F1"), ("auroc", "AUROC")]):
    for i, model in enumerate(ORDER):
        vals = df.loc[df["model"] == model, col]
        std = vals.std() if len(vals) > 1 else 0.0
        ax.bar(i, vals.mean(), yerr=std, capsize=5, color=color_of(model), alpha=0.85)
        ax.scatter(i + rng.uniform(-0.12, 0.12, len(vals)), vals, color="black", s=14, zorder=3)
    ax.set_xticks(range(len(ORDER)))
    ax.set_xticklabels(ORDER, rotation=15)
    ax.set_ylim(max(0.0, df[col].min() - 0.03), 1.005)
    ax.set_ylabel(label)
    ax.set_title(f"{label} on the held-out test set")
fig.suptitle("Mean ± std over seeds (dots = individual seeds)")
fig.tight_layout()
plt.show()
```

## 3. Confusion matrices

Counts are summed over all seeds (each seed uses its own stratified split) and each row is shown as a percentage of the true class.

```python
fig, axes = plt.subplots(1, len(ORDER), figsize=(4 * len(ORDER), 3.9), squeeze=False)
idx = [ABNORMAL_LABEL, NORMAL_LABEL]
tick_labels = ["Abnormal", "Normal"]
for ax, model in zip(axes[0], ORDER):
    total = np.sum(df.loc[df["model"] == model, "cm"].tolist(), axis=0)
    counts = total[np.ix_(idx, idx)]
    pct = counts / counts.sum(axis=1, keepdims=True) * 100
    ax.imshow(pct, vmin=0, vmax=100, cmap="Blues")
    for i in range(2):
        for j in range(2):
            ax.text(j, i, f"{pct[i, j]:.1f}%\n(n={counts[i, j]})", ha="center", va="center",
                    color="white" if pct[i, j] > 60 else "black")
    ax.set_xticks([0, 1]); ax.set_xticklabels(tick_labels)
    ax.set_yticks([0, 1]); ax.set_yticklabels(tick_labels)
    ax.set_xlabel("Predicted"); ax.set_ylabel("True")
    ax.set_title(model)
    ax.grid(False)
fig.tight_layout()
plt.show()
```

## 4. Training curves

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 4))
for model in ORDER:
    first = True
    for _, run in df[df["model"] == model].iterrows():
        h = pd.DataFrame(run["history"])
        if h.empty:
            continue
        label = model if first else None
        axes[0].plot(h["epoch"], h["loss"], color=color_of(model), alpha=0.6, label=label)
        axes[1].plot(h["epoch"], h["val_macro_f1"], color=color_of(model), alpha=0.6, label=label)
        first = False
axes[0].set_title("Training loss (log scale)")
axes[0].set_yscale("log")
axes[1].set_title("Validation macro F1 (axis zoomed to 0.9-1.0)")
axes[1].set_ylim(0.9, 1.0)
for ax in axes:
    ax.set_xlabel("Epoch")
    ax.legend()
fig.tight_layout()
plt.show()
```

## 5. Model size and cost

Inference times are wall-clock times of a **software simulation** on the device listed above. The quantum circuit is simulated on a classical computer, so its time says nothing about quantum hardware, and the spiking networks run as ordinary dense PyTorch code here, so they get no benefit from spike sparsity on this hardware. Treat this panel as a cost comparison for this code, not an edge-device energy claim.

```python
means = df.groupby("model")[["params", "ms_per_sample", "macro_f1"]].mean().loc[ORDER]
colors = [color_of(m) for m in ORDER]

fig, axes = plt.subplots(1, 3, figsize=(14, 4))
axes[0].bar(ORDER, means["params"], color=colors)
axes[0].set_title("Trainable parameters")
axes[1].bar(ORDER, means["ms_per_sample"], color=colors)
axes[1].set_yscale("log")
axes[1].set_title("Inference time (ms/sample, log scale)")
axes[2].scatter(means["params"], means["macro_f1"], s=130, c=colors, zorder=3)
for model, row in means.iterrows():
    axes[2].annotate(model, (row["params"], row["macro_f1"]), textcoords="offset points", xytext=(6, 6))
axes[2].set_xlabel("Parameters")
axes[2].set_ylabel("Mean macro F1")
axes[2].set_title("Accuracy vs. size")
for ax in axes[:2]:
    ax.tick_params(axis="x", rotation=15)
fig.tight_layout()
plt.show()
```

## 6. The data: example heartbeats
This cell assumes the label is the last column; use it to confirm which label is the normal class and update `NORMAL_LABEL` at the top if needed.

```python
data_path = Path("data/ecg.csv")
if not data_path.exists():
    print("data/ecg.csv not found: upload it to see example heartbeats.")
else:
    raw = pd.read_csv(data_path, header=None)
    if pd.to_numeric(raw.iloc[0], errors="coerce").isna().any():
        raw = pd.read_csv(data_path)  # the first row was a header
    X = raw.iloc[:, :-1].to_numpy(dtype=float)
    y = raw.iloc[:, -1].astype(int).to_numpy()
    print("Class counts (label: count, share):")
    for label, count in pd.Series(y).value_counts().sort_index().items():
        print(f"  {label}: {count} ({count / len(y):.1%})")

    fig, axes = plt.subplots(1, 2, figsize=(11, 3.8), sharey=True)
    for ax, (label, title) in zip(axes, [(ABNORMAL_LABEL, "Abnormal"), (NORMAL_LABEL, "Normal")]):
        rows_c = X[y == label]
        mean, std = rows_c.mean(axis=0), rows_c.std(axis=0)
        t = np.arange(X.shape[1])
        for example in rows_c[:5]:
            ax.plot(t, example, color="grey", alpha=0.35, lw=0.8)
        ax.plot(t, mean, color="crimson" if label == ABNORMAL_LABEL else "navy", lw=2, label="class mean")
        ax.fill_between(t, mean - std, mean + std, alpha=0.2, label="± 1 std")
        ax.set_title(f"{title} (CSV label {label}, n={len(rows_c)})")
        ax.set_xlabel("Sample index within the heartbeat")
        ax.legend()
    axes[0].set_ylabel("Amplitude (as stored in the CSV)")
    fig.tight_layout()
    plt.show()
```

**Limitations**

- One public dataset with a single recording source, so results may not transfer to other patients or devices.
- All models score near the ceiling, so small differences are within seed noise.
- The quantum circuit is simulated on a classical CPU with a handful of qubits; timings are not quantum-hardware speed.
- Spike sparsity and a real energy estimate are not measured here, so no energy-efficiency claim is made.
